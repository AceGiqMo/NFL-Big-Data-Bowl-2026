# NFL Big Data Bowl 2026 — Exploratory Data Analysis

## Goal

The goal of this notebook is to understand the structure of the NFL
player-tracking data and find patterns that can help us build a model
for predicting future player trajectories.

We focus on:

- the structure of `input` and `output`;
- players, plays and frames;
- player movement;
- speed, acceleration and direction;
- ball landing position;
- target players;
- prediction horizon;
- relationships between the input features and future movement.

The final section contains the main findings that can be used during
feature engineering and model development.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from PIL import Image

from pathlib import Path

image_list = []

pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", 100)

sns.set_theme(style="whitegrid")

DATA_PATH = Path("data")

# Keep this as ["01"] while developing locally.
# Change to None for all available input_2023_wXX.csv files.
WEEKS_TO_LOAD = ["01"]
```

## 1. Imports and configuration

For development, the notebook uses the first available week. After the
EDA is finished, `WEEKS_TO_LOAD` can be changed to `None` to load all
available training weeks.

## 2. Loading the data

```python
input_files = sorted(DATA_PATH.glob("input_2023_w*.csv"))
output_files = sorted(DATA_PATH.glob("output_2023_w*.csv"))

print(f"Input files found: {len(input_files)}")
print(f"Output files found: {len(output_files)}")

if not input_files:
    raise FileNotFoundError(f"No input files found in {DATA_PATH}")

if WEEKS_TO_LOAD is not None:
    wanted = {f"{int(w):02d}" for w in WEEKS_TO_LOAD}
    input_files = [
        f for f in input_files
        if f.stem[-2:] in wanted
    ]
    output_files = [
        f for f in output_files
        if f.stem[-2:] in wanted
    ]

print("Files used:")
for f in input_files:
    print(" ", f.name)
```

```python
input_parts = []
output_parts = []

for file in input_files:
    df = pd.read_csv(file)
    df["week"] = file.stem[-2:]
    input_parts.append(df)

for file in output_files:
    df = pd.read_csv(file)
    df["week"] = file.stem[-2:]
    output_parts.append(df)

input_data = pd.concat(input_parts, ignore_index=True)
output_data = pd.concat(output_parts, ignore_index=True)

print("Input shape:", input_data.shape)
print("Output shape:", output_data.shape)
```

## 3. Dataset structure

The input contains the information available before the prediction
period. The output contains the future `x` and `y` coordinates that the
model has to predict.

```python
print("INPUT COLUMNS")
for column in input_data.columns:
    print("-", column)

print("\nOUTPUT COLUMNS")
for column in output_data.columns:
    print("-", column)
```

```python
summary = pd.DataFrame({
    "dataset": ["input", "output"],
    "rows": [len(input_data), len(output_data)],
    "columns": [input_data.shape[1], output_data.shape[1]],
    "games": [
        input_data["game_id"].nunique(),
        output_data["game_id"].nunique()
    ],
    "plays": [
        input_data[["game_id", "play_id"]].drop_duplicates().shape[0],
        output_data[["game_id", "play_id"]].drop_duplicates().shape[0]
    ],
    "players": [
        input_data["nfl_id"].nunique(),
        output_data["nfl_id"].nunique()
    ]
})

display(summary)
```

### Key observation

A play is identified by the combination of `game_id` and `play_id`.
The output has only six main columns: the identifiers, frame number,
and the future `x` and `y` coordinates.

This means that the prediction problem can be viewed as forecasting a
future sequence of `(x, y)` positions for selected players.

## 4. Missing values

```python
missing = input_data.isna().sum().sort_values(ascending=False)
missing = missing[missing > 0]

if missing.empty:
    print("No missing values were found in the input data.")
else:
    display(missing.to_frame("missing_values"))

missing_output = output_data.isna().sum().sort_values(ascending=False)
missing_output = missing_output[missing_output > 0]

if missing_output.empty:
    print("No missing values were found in the output data.")
else:
    display(missing_output.to_frame("missing_values"))
```

### Key observation

Missing values should be checked before modeling. In particular,
dynamic tracking features and static player information should be
handled separately if missing values appear in the full dataset.

## 11. Distance to the ball landing position

We create a derived feature measuring the Euclidean distance from the
player's current position to the expected ball landing point.

```python
dx = input_data["ball_land_x"] - input_data["x"]
dy = input_data["ball_land_y"] - input_data["y"]

input_data["distance_to_ball"] = np.sqrt(dx**2 + dy**2)

display(input_data["distance_to_ball"].describe())
```

```python
plt.figure(figsize=(10, 5))
sns.histplot(
    data=input_data,
    x="distance_to_ball",
    bins=50
)
plt.title("Distance to Ball Landing Position")
plt.xlabel("Distance (yards)")
plt.ylabel("Count")
plt.savefig('plot9.png', dpi=300, bbox_inches='tight')
image_list.append('plot9.png')
plt.show()
```

## 14. Input trajectory vs future trajectory

The next plot shows the actual future trajectory from `output` together
with the observed trajectory from `input`.

This is the clearest visual representation of the machine-learning task:
the solid part is available to the model, while the dashed part is what
the model needs to predict.

```python
plt.figure(figsize=(14, 7))

target_ids = set(
    play_input.loc[play_input["player_to_predict"], "nfl_id"]
)

for nfl_id, player in play_input.groupby("nfl_id"):
    is_target = nfl_id in target_ids

    plt.plot(
        player["x"],
        player["y"],
        linewidth=3 if is_target else 1,
        alpha=1.0 if is_target else 0.12
    )

for nfl_id, player in play_output.groupby("nfl_id"):
    plt.plot(
        player["x"],
        player["y"],
        linestyle="--",
        linewidth=2.5
    )

plt.scatter(
    play_input["ball_land_x"].iloc[0],
    play_input["ball_land_y"].iloc[0],
    marker="*",
    s=300,
    label="Ball landing"
)

plt.xlabel("X position (yards)")
plt.ylabel("Y position (yards)")
plt.title("Input and Future Player Trajectories")
plt.show()
```

### Key observation

Future movement is not simply a straight movement toward the ball
landing point. Players can change direction after the throw.

Therefore, a useful model should combine the player's movement history
with contextual information such as the ball landing position and the
positions and movements of other players.

## 15. Movement during the input sequence

For a target player, it is useful to measure how much the player moves
during the observed history.

```python
target_input = input_data[input_data["player_to_predict"]].copy()

target_movement = (
    target_input
    .sort_values(["game_id", "play_id", "nfl_id", "frame_id"])
    .groupby(["game_id", "play_id", "nfl_id"])
    .agg(
        start_x=("x", "first"),
        start_y=("y", "first"),
        end_x=("x", "last"),
        end_y=("y", "last"),
        mean_speed=("s", "mean"),
        max_speed=("s", "max"),
        mean_acceleration=("a", "mean"),
        frames=("frame_id", "count")
    )
    .reset_index()
)

target_movement["observed_displacement"] = np.sqrt(
    (target_movement["end_x"] - target_movement["start_x"]) ** 2 +
    (target_movement["end_y"] - target_movement["start_y"]) ** 2
)

display(target_movement.describe())
```

```python
plt.figure(figsize=(10, 6))
sns.scatterplot(
    data=target_movement.sample(
        min(30000, len(target_movement)),
        random_state=42
    ),
    x="mean_speed",
    y="observed_displacement",
    alpha=0.25
)
plt.title("Mean Speed vs Observed Displacement of Target Players")
plt.xlabel("Mean speed")
plt.ylabel("Observed displacement (yards)")
plt.savefig('plot13.png', dpi=300, bbox_inches='tight')
image_list.append('plot13.png')
plt.show()
```

### Key observation

The observed movement history contains information about how actively a
target player is moving before the prediction period. Features based on
recent speed, acceleration and displacement may therefore help predict
future positions.

# 17. Key Findings for Modeling

The EDA suggests several important directions for the future model.

### 1. Data structure: sequence-to-sequence task, no missing values

- `input`: 285,714 rows × 24 columns — tracking data for all relevant players
  **before the throw**.
- `output`: 32,088 rows × 7 columns — **future (x, y) coordinates** only for
  players that must be evaluated.
- 819 passes and 16 games are included in week 01.
- There are 737 unique players in `input` and 563 unique players in `output`.
  This confirms that `output` contains only players with
  `player_to_predict = True`; all other players are used as context.
- No missing values were found in either table. Therefore, imputation is not
  necessary, but masks for variable-length sequences `(S, O, P)` are still
  required.

| Quantity (week 01) | Value |
|---|---|
| input / output rows × columns | 285,714 × 24 / 32,088 × 7 |
| games / passes | 16 / 819 |
| unique players in input / output | 737 / 563 |
| players per pass | 6–14, median 13, mean 12.3 |
| target players per pass (`player_to_predict`) | 1–8, median 3, mean 3.3 |
| input history length S | 9–74 frames, median 27, mean 28.5 |
| output horizon O = `num_frames_output` | 5–94 frames, median 10, mean 11.1 |
| speed `s` | mean 2.88 yd/s, max 12.5 |
| acceleration `a` | mean 2.09 yd/s² |
| displacement during input history | mean 7.6 yd, std 4.9, max 31.7 |
| play direction | right 53.3% / left 46.7% |
| distance to ball landing point | mean 19.1 yd, std 10.6, max 65.7 |
| missing values | none |

### 2. Variable number of agents and player roles

- The number of players per pass ranges from 6 to 14, with a median of 13.
  Therefore, `P` is **not constant**, and padding plus masks are required in all
  spatial layers.
- Since `P ≤ 14`, spatial attention is computationally cheap. This means that
  deeper spatial layers or pairwise interaction features can be considered
  without a large computational cost.
- The number of target players per pass ranges from 1 to 8, with a median of 3.
  Therefore, the model has a multi-agent output with a target mask.
- Crosstab analysis of `player_role × player_to_predict` shows:
  - **Targeted Receiver** is always evaluated.
  - **Defensive Coverage** is evaluated for a subset of defenders.
  - **Passer** and **Other Route Runner** are never evaluated and act only as
    contextual players.
- Implications:
  - The loss must be masked by `player_to_predict` — this is already implemented.
  - Player role should be used as a conditional feature or embedding.
  - Metrics should be reported by role — this is already implemented.
  - Since there are more evaluated defenders than targeted receivers, a
    role-weighted loss is a possible future improvement.

### 3. Variable input/output lengths and a heavy-tailed prediction horizon

- Input history length `S` ranges from 9 to 74 frames per player, with median 27
  and mean 28.5 at 10 Hz.
- This supports the need for:
  - sequence masks;
  - a recency/time-decay feature;
  - a model that can process variable-length histories.
- The current cap `max_input_frames = 64` cuts off the longest histories
  observed in week 01, where the maximum is 74. It should be increased to 80.
- Output horizon `O` ranges from 5 to 94 frames, with median 10 and mean 11.1.
  The distribution has a strong right tail: most passes are short, but some
  long passes require forecasting up to 94 frames, i.e. 9.4 seconds.
- **Important:** the official metric scores all output frames. Therefore,
  `max_output_frames = 48` cuts off frames that are part of the evaluation.
  It should be increased to 96.
- Long horizons are especially difficult because delta predictions are
  accumulated with `cumsum`, so small per-frame errors drift over time.
- Possible improvements:
  - auxiliary loss on absolute positions;
  - horizon-weighted loss;
  - residual prediction relative to a simple kinematic baseline.

```python
# Config adjustment based on EDA.
# The statistics are from week 01, so the caps are chosen with a safety margin.

cfg.max_input_frames = 80    # observed S: 9–74; the previous cap 64 truncated long histories
cfg.max_output_frames = 96   # observed O: 5–94; the previous cap 48 truncated scored frames

# P is handled by padding to the maximum number of players inside each batch.
# The observed maximum in week 01 is 14, and the existing collate function already supports this.

cfg.huber_delta = 0.35       # mean per-frame displacement ≈ 2.88 yd/s / 10 Hz ≈ 0.29 yd
```

### 4. Kinematics: feature scales and motivation for Huber loss

- Mean speed is 2.88 yd/s, with a maximum of 12.5 yd/s.
- Mean acceleration is 2.09 yd/s².
- The scatter plot of average speed versus observed displacement shows a clear
  monotonic relationship, so kinematic features are informative.
- Mean displacement during the observed input window is 7.6 yards, with standard
  deviation 4.9 and maximum 31.7 yards.
- This means that players often move substantially **before** the throw, so the
  full pre-pass trajectory contains useful information. A single release-frame
  snapshot is not enough.
- At 10 Hz, the average per-frame displacement is approximately 0.29 yards.
  Therefore, `huber_delta = 0.35` is a reasonable scale:
  - small typical errors remain in the quadratic part of Huber loss;
  - large errors from sharp cuts or unusual reactions are handled by the linear
    robust part.
- The constant-velocity baseline obtains RMSE ≈ 4.25 yards, while the trained
  model obtains RMSE ≈ 0.80 yards. This shows that pure kinematic extrapolation
  is not sufficient: after the throw, players change direction and react to the
  ball and to other players.

### 5. Circular angles and field orientation

- `dir` and `o` are angular variables in degrees, ranging from 0° to 360°.
- Values close to 0° and 360° represent almost the same direction, but they are
  numerically far apart if treated as ordinary linear variables.
- Therefore, direct Pearson correlation on these variables is not very
  meaningful.
- The current model handles this correctly:
  - `dir` and `o` are encoded as `sin(angle)` and `cos(angle)`;
  - plays moving left are flipped so that the offense always attacks in the
    positive x-direction.
- Play direction is almost balanced: right 53.3%, left 46.7%. Normalizing field
  orientation makes similar football situations more comparable.

### 6. Ball landing point: strong geometric anchor and role-dependent context

- The distribution of ball landing positions covers the whole field.
- Mean landing position is approximately `(60.9, 26.2)`.
- Some landing coordinates are slightly outside the nominal field boundaries
  (`y` ranges from about −1.7 to 57.3). Therefore, landing coordinates should
  not be hard-clipped.
- `distance_to_ball` has mean 19.1 yards, standard deviation 10.6, and maximum
  65.7.
- Boxplots by side and role show substantial differences:
  - the passer is usually far from the landing point;
  - targeted receivers and coverage defenders are closer and more variable.
- This means that distance to the ball landing point is a role-dependent and
  informative feature.
- Correlation analysis shows:
  - `x` and `ball_land_x`: correlation ≈ 0.88;
  - `distance_to_ball` and output horizon: correlation ≈ 0.52.
- Interpretation:
  - the landing point strongly anchors player positions, especially downfield;
  - deeper passes tend to have longer flight times and larger player-to-ball
    distances.
- The current architecture already uses static distance-to-ball features and
  horizon progress features.
- Possible improvement: add direction-to-landing features, such as normalized
  vector from the player to the ball landing point and radial velocity toward
  the landing point.

### 7. Correlation matrix: useful, but insufficient

- Most pairwise linear correlations are small: typically `|r| ≤ 0.25`.
- Important exceptions:
  - `x` and `ball_land_x`: ≈ 0.88;
  - `distance_to_ball` and prediction horizon: ≈ 0.52;
  - `s` and `a`: ≈ 0.21.
- This does **not** mean that the remaining features are useless.
- The data is sequential, multi-agent and nonlinear; useful information can
  exist in temporal patterns and interactions even when pairwise correlations
  are weak.
- Angular variables are circular, so their raw linear correlations are especially
  hard to interpret.
- Conclusion:
  - do not perform feature selection only by linear correlation;
  - keep the full kinematic feature set;
  - let the sequence model learn nonlinear temporal and interaction patterns.

### 8. Single-play visualization: future movement is not a straight line

- The plot comparing input trajectories with future trajectories shows that
  future movement is often curved rather than linear.
- Players make route breaks, slow down, accelerate, or react to the ball and to
  other players.
- Therefore, a straight-line extrapolation toward the ball landing point is not
  sufficient.
- This supports the current modeling strategy:
  - temporal modeling of pre-pass movement;
  - spatial attention over nearby players;
  - explicit ball landing features;
  - delta prediction followed by cumulative summation.
- The same plot also explains a limitation of delta prediction:
  local motion is easy to model, but small errors accumulate over long horizons.

### 9. Data splitting and leakage

- The notebook explicitly warns that rows from the same play must not be split
  randomly between training and validation.
- If rows from the same play appear in both train and validation, the model can
  indirectly see parts of the same trajectory during training, which creates
  leakage and over-optimistic validation metrics.
- The current project uses a week-based split:
  - train: weeks 01–15;
  - validation: weeks 16–17;
  - test: week 18.
- This split prevents play-level leakage and is stricter than a random row-level
  split.
- Reminder: the EDA was performed on week 01 only. Therefore, the observed
  ranges are development estimates; caps such as 80/96 should be used with
  margin for the full 18-week dataset.

### 10. Summary: EDA observation → architectural decision

| EDA observation | Architectural decision | Status |
|---|---|---|
| `P` varies from 6 to 14 | padding + masks in all spatial layers; pad to max `P` inside each batch | already implemented |
| `S` varies from 9 to 74 | sequence masks, time-decay feature; increase cap from 64 to **80** | adjustment |
| `O` varies from 5 to 94 with a long right tail | RoPE over horizon, progress feature, masks; increase cap from 48 to **96** | adjustment |
| only a subset of players is evaluated | loss mask by `player_to_predict`; role-level metrics | already implemented |
| player roles differ strongly | possible role embedding and role-weighted loss | future improvement |
| angles are circular | `sin/cos` encoding for `dir` and `o` | already implemented |
| play direction is roughly balanced | normalize all plays so offense attacks +x | already implemented |
| ball landing point is a strong anchor | use distance-to-landing features; possible direction-to-landing features | partially implemented |
| mean per-frame displacement ≈ 0.29 yd | Huber loss with `delta = 0.35` | already implemented |
| long-horizon cumsum drift exists | possible auxiliary absolute-position loss or horizon-weighted loss | future improvement |
| random row split causes leakage | use week-based split | already implemented |

# Conclusion

The EDA confirms that the task is naturally a masked sequence-to-sequence,
multi-agent trajectory prediction problem. Each play contains a variable number
of players, variable input history length and variable output horizon. Only a
subset of players is evaluated, while the remaining players provide important
context.

The current architecture is well aligned with these findings:

- delta prediction with Huber loss matches the scale and heavy-tailed nature of
  per-frame movement;
- sine/cosine angle encoding correctly handles circular features;
- play-direction normalization makes comparable situations share one coordinate
  system;
- temporal blocks model pre-pass movement history;
- spatial attention models interactions between players;
- ball landing features provide a strong geometric anchor;
- RoPE and explicit progress features condition the decoder on prediction
  horizon;
- week-based splitting avoids play-level leakage.

The main configuration changes suggested by EDA are:

- increase `max_input_frames` from 64 to **80**;
- increase `max_output_frames` from 48 to **96**.

The most promising future improvements are:

1. auxiliary loss on absolute positions to reduce long-horizon cumsum drift;
2. role embedding and/or role-weighted loss;
3. direction-to-ball-landing features;
4. horizon-weighted loss for long passes;
5. controlled ablations for Muon and mixed precision after stability guards are
   in place.

Overall, the EDA supports the existing spatio-temporal design and also gives
clear directions for the next stage: custom architecture improvements and
hyperparameter tuning.
