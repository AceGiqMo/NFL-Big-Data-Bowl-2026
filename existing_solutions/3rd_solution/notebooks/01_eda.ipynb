# NFL Big Data Bowl 2026 — Small EDA

Model-oriented EDA for reproducing the 3rd-place solution. Run on one week first.

```python
from pathlib import Path
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

DATA_DIR = Path('../data/train')
WEEK = 1
inp = pd.read_csv(DATA_DIR / f'input_2023_w{WEEK:02d}.csv')
out = pd.read_csv(DATA_DIR / f'output_2023_w{WEEK:02d}.csv')
print(inp.shape, out.shape)
```

```python
inp.groupby(['game_id','play_id'])['nfl_id'].nunique().describe()
```

```python
inp.groupby(['game_id','play_id'])['num_frames_output'].first().describe()
```

```python
fig, ax = plt.subplots(figsize=(7,4))
inp['s'].hist(bins=50, ax=ax)
ax.set_title('Player speed')
ax.set_xlabel('yards / second')
plt.show()
```

```python
roles = inp['player_role'].value_counts()
roles.plot(kind='bar', figsize=(8,4), title='Player roles in input tracking data')
plt.ylabel('rows')
plt.show()
```

## Key ML observations

- Data are sequences of tracking frames rather than iid rows.
- Up to 22 player trajectories coexist in one play, so player interactions matter.
- `s`, `a`, `dir`, `o` provide motion state.
- `ball_land_x/y` provide future ball context available before prediction.
- The prediction horizon varies by play/player, so masking is necessary.
- The 3rd-place solution therefore uses fixed player slots + temporal and spatial attention.
